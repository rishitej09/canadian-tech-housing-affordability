```python
#### CODE TO RUN ONCE IT'S TIME

import requests
import pandas as pd
import math
import time
from datetime import datetime, timezone


# ============================================================
# 1. API CONFIGURATION
# ============================================================

#APP_ID = "YOUR_APP_ID"
#APP_KEY = "YOUR_APP_KEY"

APP_ID = "4203ae48"
APP_KEY = "9b933e533406683028e19af97ae403bf"

COUNTRY = "ca"

headers = {
    "Accept": "application/json"
}


# ============================================================
# 2. RESEARCH PARAMETERS
# ============================================================

positions = [
    "Data Analyst",
    "Data Scientist",
    "Data Engineer"
]

cities = [
    "Toronto",
    "Vancouver",
    "Montreal"
]

RESULTS_PER_PAGE = 50
MAX_DAYS_OLD = 280
DISTANCE = 26

REQUEST_DELAY = 2


# ============================================================
# 3. STORAGE
# ============================================================

all_jobs = []


# ============================================================
# 4. LOOP THROUGH ALL POSITION-CITY COMBINATIONS
# ============================================================

for position in positions:

    for city in cities:

        print("\n" + "=" * 70)
        print(f"Starting search: {position} | {city}")
        print("=" * 70)

        # ----------------------------------------------------
        # FIRST REQUEST
        # ----------------------------------------------------

        page = 1

        URL = (
            f"https://api.adzuna.com/v1/api/jobs/"
            f"{COUNTRY}/search/{page}"
        )

        params = {
            "app_id": APP_ID,
            "app_key": APP_KEY,
            "results_per_page": RESULTS_PER_PAGE,
            "what": position,
            "where": city,
            "content-type": "application/json",
            "max_days_old": MAX_DAYS_OLD,
            "distance": DISTANCE
        }

        try:

            response = requests.get(
                URL,
                params=params,
                headers=headers,
                timeout=30
            )

            response.raise_for_status()

            data = response.json()

            total_jobs = data.get("count", 0)
            jobs = data.get("results", [])

            print("Total jobs found:", total_jobs)
            print("Jobs returned on page 1:", len(jobs))

            # Calculate only the pages actually needed
            total_pages = math.ceil(
                total_jobs / RESULTS_PER_PAGE
            )

            print("Total pages needed:", total_pages)

            # Save page 1
            request_datetime = datetime.now(
                timezone.utc
            ).isoformat()

            for job in jobs:

                job["search_position"] = position
                job["search_city"] = city
                job["request_datetime"] = request_datetime
                job["source_site"] = "Adzuna"

                all_jobs.append(job)

            # If there are no more pages, move to next combination
            if total_pages <= 1:
                continue

            time.sleep(REQUEST_DELAY)


            # ------------------------------------------------
            # REQUEST REMAINING PAGES
            # ------------------------------------------------
###########################################################################
            for page in range(2, total_pages + 1):

                URL = (
                    f"https://api.adzuna.com/v1/api/jobs/"
                    f"{COUNTRY}/search/{page}"
                )

                params = {
                    "app_id": APP_ID,
                    "app_key": APP_KEY,
                    "results_per_page": RESULTS_PER_PAGE,
                    "what": position,
                    "where": city,
                    "content-type": "application/json",
                    "max_days_old": MAX_DAYS_OLD,
                    "distance": DISTANCE
                }
                try:

                    response = requests.get(
                        URL,
                        params=params,
                        headers=headers,
                        timeout=30
                    )

                    response.raise_for_status()

                    page_data = response.json()

                    jobs = page_data.get("results", [])

                    print(
                        f"{position} | {city} | "
                        f"Page {page} | "
                        f"Jobs retrieved: {len(jobs)}"
                    )

                    # Extra safety
                    if len(jobs) == 0:
                        print(
                            "No more jobs returned. "
                            "Stopping this combination."
                        )
                        break

                    request_datetime = datetime.now(
                        timezone.utc
                    ).isoformat()

                    for job in jobs:

                        job["search_position"] = position
                        job["search_city"] = city#                        job["request_datetime"] = request_datetime
                        job["source_site"] = "Adzuna"

                        all_jobs.append(job)

                    time.sleep(REQUEST_DELAY)

                except requests.exceptions.RequestException as e:

                    print(
                        f"API error for {position} | "
                        f"{city} | Page {page}: {e}"
                    )

                    break

        except requests.exceptions.RequestException as e:

            print(
                f"Initial API request failed for "
                f"{position} | {city}: {e}"
            )

            continue


# ============================================================
# 5. CREATE DATAFRAME
# ============================================================

df_adzuna = pd.json_normalize(all_jobs)


# ============================================================
# 6. SUMMARY
# ============================================================

print("\n" + "=" * 70)
print("COLLECTION COMPLETE")
print("=" * 70)

print("Total jobs collected:", len(df_adzuna))
print("Total columns:", len(df_adzuna.columns))

if not df_adzuna.empty:

    print("\nJobs collected by position and city:")

    summary = (
        df_adzuna
        .groupby(
            ["search_position", "search_city"]
        )
        .size()
        .reset_index(name="jobs_collected")
    )

    display(summary)

    print("\nPreview:")
    display(df_adzuna.head())


# ============================================================
# 7. SAVE CSV
# ============================================================

output_file = "adzuna_raw_jobs.csv"

df_adzuna.to_csv(
    output_file,
    index=False,
    encoding="utf-8-sig"
)

print(f"\nCSV saved successfully: {output_file}")
```
